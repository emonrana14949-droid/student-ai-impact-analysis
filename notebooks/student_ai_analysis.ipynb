# Student AI Impact Analysis

A Pandas, NumPy and Matplotlib walkthrough of the student AI dataset.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')
df = pd.read_csv('../data/student_ai_dataset.csv')
df.head()
```

## 1. Average score by AI tool


```python
by_tool = df.groupby('ai_tool').agg(
    students=('student_id', 'count'),
    avg_ai_hours=('ai_usage_hours_per_week', 'mean'),
    avg_creativity_score=('creativity_score', 'mean'),
    avg_final_score=('final_score', 'mean')
).round(2).sort_values('avg_final_score', ascending=False)
by_tool
```

## 2. Correlation check


```python
cols = ['ai_usage_hours_per_week', 'study_hours_per_week', 'attendance_percentage', 'creativity_score', 'final_score']
df[cols].corr(numeric_only=True).round(2)
```

## 3. Visualization: AI usage vs final score


```python
slope, intercept = np.polyfit(df['ai_usage_hours_per_week'], df['final_score'], 1)
x = np.linspace(df['ai_usage_hours_per_week'].min(), df['ai_usage_hours_per_week'].max(), 100)

plt.figure(figsize=(9, 5))
plt.scatter(df['ai_usage_hours_per_week'], df['final_score'], alpha=0.65)
plt.plot(x, slope * x + intercept, color='black', linewidth=1.5)
plt.title('AI Usage Hours vs Final Score')
plt.xlabel('AI usage hours per week')
plt.ylabel('Final score')
plt.show()
```
